```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np

# =================================================================
                                
# =================================================================
class FourierFeatureMapping(nn.Module):
    def __init__(self, in_dims=2, mapping_size=128, scale=10.0):
        super().__init__()
                                  
        self.register_buffer('B', torch.randn(in_dims, mapping_size) * scale)

    def forward(self, x):
        # x: (Batch, 2)
        # x @ B: (Batch, mapping_size)
        x_proj = (2.0 * np.pi * x) @ self.B
                                          
        return torch.cat([torch.sin(x_proj), torch.cos(x_proj)], dim=-1)

# =================================================================
               
# =================================================================
class DenseResBlock(nn.Module):
    def __init__(self, size):
        super().__init__()
        self.fc1 = nn.Linear(size, size)
        self.fc2 = nn.Linear(size, size)
        self.norm1 = nn.LayerNorm(size)
        self.norm2 = nn.LayerNorm(size)

    def forward(self, x):
        res = x
        x = self.norm1(x)
        x = F.silu(self.fc1(x))
        x = self.norm2(x)
        x = F.silu(self.fc2(x))
        return x + res

# =================================================================
                       
# =================================================================
class HighFidelityNeuralVF(nn.Module):
    def __init__(self, n_poles=8, width=256, fourier_size=128,num=6):
        super().__init__()
        self.n_poles = n_poles
        
                                         
        self.fourier_map = FourierFeatureMapping(in_dims=2, mapping_size=fourier_size, scale=5.0)
        
                                 
        in_features = fourier_size * 2 + 2 
        self.in_layer = nn.Linear(in_features, width)
        
                           
        self.res_layers = nn.ModuleList([DenseResBlock(width) for _ in range(num)])
        
                
        self.out_layer = nn.Linear(width, 2 * n_poles)

    def forward(self, x):
        # x shape: (Batch, 2)
        
                
        x_fourier = self.fourier_map(x)
                                            
        x_concat = torch.cat([x, x_fourier], dim=-1)
        
        h = F.silu(self.in_layer(x_concat))
        for layer in self.res_layers:
            h = layer(h)
        out = self.out_layer(h)
        
                             
        log_tau_k = out[:, 0 : self.n_poles]
        log_w_k = out[:, self.n_poles :]
        
        tau_k = 10.0 ** log_tau_k
        w_k = 10.0 ** log_w_k
                                  
                                        
        w_k_sum = torch.sum(w_k, dim=1, keepdim=True) + 1e-12        
        w_k = w_k / w_k_sum
        
        return tau_k, w_k, log_tau_k, log_w_k
```

```python
def physics_informed_loss( p_tau, p_w, p_log_tau, p_log_w, inputs, s_bench):

                                    
    log_t_cc = inputs[:, 0:1] 
    c_cc = inputs[:, 1:2]     
    t_cc = 10 ** log_t_cc     

    s_tensor = s_bench.view(1, -1) 
    y_true = 1.0 / (1.0 + (s_tensor * t_cc) ** c_cc)
    
    s_3d = s_bench.view(1, -1, 1)
    tau_3d = p_tau.unsqueeze(1)
    w_3d = p_w.unsqueeze(1)
    
    denom = 1.0 + s_3d * tau_3d
    terms = w_3d / denom        
    y_pred = torch.sum(terms, dim=-1)
    
    loss_phys = torch.mean(torch.abs((y_true - y_pred))**2)
    
    return loss_phys
```

```python
n_samples=10000
np.random.seed(42)
torch.manual_seed(42)

tau_samples = np.random.uniform(-4,1.0, n_samples)
c_samples = np.random.uniform(0.1, 1.0, n_samples)

X_np = np.stack([tau_samples, c_samples], axis=1)
X_np = np.array(X_np)
```

```python

# =================================================================
              
# =================================================================

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")


                   
dataset = TensorDataset(
    torch.tensor(X_np, dtype=torch.float32).to(device),
    torch.tensor(X_np, dtype=torch.float32).to(device),

)
dataloader = DataLoader(dataset, batch_size=512, shuffle=True)

```

```python
              
model = HighFidelityNeuralVF(n_poles=14, width=512, fourier_size=64,num=4).to(device)

epochs = 10000

optimizer = optim.Adam(model.parameters(), lr=1e-4)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=epochs, eta_min=1e-6)

```

```python
                                           
freqs_bench = torch.logspace(-2, 6, 100).to(device)
s_bench = 1j * 2 * torch.pi * freqs_bench
```

```python

                        
import copy

early_stop_patience = 500
best_loss = float('inf')
best_model_wts = copy.deepcopy(model.state_dict())
epochs_no_improve = 0

print("\n[*] 开始 PyTorch 网络训练...")
model.train()
for epoch in range(epochs):
    epoch_loss = 0.0
    for batch_x, batch_d0 in dataloader:
        optimizer.zero_grad()
        
                          
        p_tau, p_w, p_log_tau, p_log_w = model(batch_x)
                   
        loss = physics_informed_loss(
             p_tau, p_w, p_log_tau, p_log_w, 
            batch_x,  s_bench
        )
        loss.backward()
        optimizer.step()

        epoch_loss += loss.item()

    avg_loss = epoch_loss / len(dataloader)
                                        
    scheduler.step()

              
    if avg_loss < best_loss:
        best_loss = avg_loss
        best_model_wts = copy.deepcopy(model.state_dict())
        epochs_no_improve = 0
    else:
        epochs_no_improve += 1

    if (epoch + 1) % 100 == 0 or epoch == 0:
        print(f"Epoch [{epoch+1}/{epochs}] | Loss: {avg_loss:.6e} | LR: {optimizer.param_groups[0]['lr']:.2e}")

        
    if epochs_no_improve >= early_stop_patience:
        print(f"[*] 满足早停条件，在第 {epoch+1} 轮停止训练。")
        break

              
model.load_state_dict(best_model_wts)

print("[*] 训练完成！(已保存最优模型权重)")
```

```python
                                   
model.eval()
num_tests = 2000

              
np.random.seed(42)

                                             
tau_samples = np.random.uniform(-3, 1, num_tests)
c_samples = np.random.uniform(0.1, 1, num_tests)
X_test = np.stack([tau_samples, c_samples], axis=1)

y_true_list, y_pred_list = [], []

for i in range(num_tests):
    tau_val = 10 ** X_test[i, 0]
    c_val = X_test[i, 1]
    x_test = torch.tensor([[np.log10(tau_val), c_val]], dtype=torch.float32).to(device)
    with torch.no_grad():
        tau_pred, w_pred, _, _ = model(x_test)
    tau_np = tau_pred.cpu().numpy()[0]
    w_np = w_pred.cpu().numpy()[0]
          
    freqs = np.logspace(-2, 6, 128)
    s = 1j * 2 * np.pi * freqs
         
    Y_true = 1.0 / (1.0 + (s * tau_val) ** c_val)
         
    Y_pred = np.sum(w_np[:, None] / (1.0 + s[None, :] * tau_np[:, None]), axis=0)
        
    y_true_list.append(Y_true)
    y_pred_list.append(Y_pred)

                            
y_true_arr = np.stack(y_true_list)
y_pred_arr = np.stack(y_pred_list)
mae = np.mean(np.abs(y_true_arr - y_pred_arr))
print(f"[*] 测试集共 {num_tests} 个点，模型输出复数均值绝对误差（MAE）：{mae:.3e}")
```
